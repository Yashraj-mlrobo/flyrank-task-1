## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import os
import pandas as pd
import numpy as np
# 1. Checking since data was not present locally
csv_path = "data/raw/content_refresh_anonymized.csv"

if not os.path.exists(csv_path):
    print("🔄 Running inside Google Colab environment. Pulling starter dataset from source...")
    # Create the folder layout inside the virtual machine
    os.makedirs("data/raw", exist_ok=True)

    # Download the exact anonymized dataset directly into the notebook's file system
    url = "https://raw.githubusercontent.com/flyrank-bih/flyrank-ml-internship-starter/main/data/raw/content_refresh_anonymized.csv"
    import urllib.request
    urllib.request.urlretrieve(url, csv_path)
    print("✅ Dataset successfully pulled into Colab workspace!")

# 2. Load the starter dataset safely
df = pd.read_csv(csv_path)

# 3. Calculate Metric 1: The Zero-Position Structural Anomaly (Gotcha Check)
unindexed_count = (df['avg_position'] == 0).sum()

# Clean out the unindexed rows to ensure our baseline stats are accurate
active_df = df[df['avg_position'] > 0].copy()

# 4. Calculate Metrics 2 & 3: The High-Stakes CTR Decay Cliff
top_tier_ctr = active_df[active_df['avg_position'] <= 0.55]['ctr'].mean()
striking_tier_ctr = active_df[(active_df['avg_position'] > 0.55) & (active_df['avg_position'] <= 1.5)]['ctr'].mean()

# 5. Output the verification numbers for your assignment submission
print("\n--- STARTER DATA VALIDATION REPORT ---")
print(f"Total Rows Evaluated: {len(df):,}")
print(f"1. Position Anomaly Quirk: {unindexed_count:,} pages contain 'avg_position = 0' (unindexed/no data).")
print(f"2. Top-Tier Performance Baseline: Pages at avg_position <= 0.55 secure a {top_tier_ctr:.2f}% mean CTR.")
print(f"3. The CTR Cliff Hazard: Dropping into the striking tier (>0.55) plunges mean CTR to {striking_tier_ctr:.2f}%.")
print("--------------------------------------")
```
